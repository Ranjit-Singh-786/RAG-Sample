```python
from langchain_community.document_loaders.csv_loader import CSVLoader
from langchain_community.vectorstores import FAISS 
from langchain_openai import OpenAIEmbeddings,ChatOpenAI
from dotenv import load_dotenv
import os 
load_dotenv()
```

```python
file_path = "organizations.csv"
loader = CSVLoader(file_path=file_path,
                   source_column="Industry",
                   metadata_columns=["Website", "Founded", "Number of employees"],
                   content_columns="Description")
```

```python
documents = loader.load()
print(documents[0])
```

```python
documents[0].page_content
```

```python
documents[0].metadata
```

#### Data Indexing

```python
emmbedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = FAISS.from_documents(documents, emmbedding_model)
```

```python
vectorstore.save_local("faiss_index")
```

```python
os.listdir(r"C:\Users\Ranjit\Desktop\RAG-Sample\RAG CSVLoader with metadata\faiss_index")
```

```python
vectordb = FAISS.load_local(
    folder_path="faiss_index",
    embeddings=emmbedding_model,
    allow_dangerous_deserialization=True
)
```

```python
result = vectordb.similarity_search("Programmable intermediate conglomeration", k=1)
```
